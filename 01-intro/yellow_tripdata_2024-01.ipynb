```python
import pandas as pd
```

```python
url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-01.parquet"
df = pd.read_parquet(url)
```

```python
print(df.shape)
df.head()
```

```python
type(df)
```

```python
df["duration"] = (df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]).dt.total_seconds() / 60
print(df["duration"].mean())
print(df["duration"].std())
```

```python
df["duration"].describe()
```

```python
n_before = len(df)
df = df[(df["duration"] >= 1) & (df["duration"] <= 60)]
n_after = len(df)

print("Avant :", n_before)
print("Après :", n_after)
print("Supprimées :", n_before - n_after)
print("Part conservée :", n_after / n_before)
```

```python
# On ne lit QUE la colonne 'trip_distance' depuis le fichier Parquet
df_reduit = pd.read_parquet(url, columns=["trip_distance"])

# Maintenant, le calcul est ultra rapide ET la RAM est préservée
moyenne = df_reduit["trip_distance"].mean()
```
